Data Visualization I (Titanic Dataset)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = sns.load_dataset('titanic')

dataset.head()
```

```python
dataset.shape
```

```python
dataset.columns
```

```python
dataset.info()
```

```python
dataset.describe(include='all')
```

```python
sns.histplot(dataset['age'],
             kde=True,
             bins=10)

plt.title("Distribution Plot of Age")

plt.show()
```

```python
sns.histplot(dataset['age'],
             kde=False,
             bins=10)

plt.title("Histogram of Age")

plt.show()
```

```python
sns.jointplot(x=dataset['age'],
              y=dataset['fare'],
              kind='scatter')

plt.show()
```

```python
sns.jointplot(x=dataset['age'],
              y=dataset['fare'],
              kind='hex')

plt.show()
```

```python
sns.rugplot(dataset['fare'])

plt.title("Rug Plot of Fare")

plt.show()
```

```python
sns.barplot(x='sex',
            y='age',
            data=dataset)

plt.title("Average Age of Male and Female Passengers")

plt.show()
```

```python
sns.barplot(x='sex',
            y='age',
            data=dataset,
            estimator=np.std)

plt.title("Standard Deviation of Age")

plt.show()
```

```python
sns.countplot(x='sex',
              data=dataset)

plt.title("Count of Male and Female Passengers")

plt.show()
```

```python
sns.boxplot(x='sex',
            y='age',
            data=dataset)

plt.title("Box Plot of Age")

plt.show()
```

```python
sns.boxplot(x='sex',
            y='age',
            data=dataset,
            hue='survived')

plt.title("Box Plot with Survival")

plt.show()
```

```python
sns.violinplot(x='sex',
               y='age',
               data=dataset)

plt.title("Violin Plot")

plt.show()
```

```python
sns.violinplot(x='sex',
               y='age',
               data=dataset,
               hue='survived')

plt.title("Violin Plot with Survival")

plt.show()
```

```python
sns.stripplot(x='sex',
              y='age',
              data=dataset,
              jitter=False)

plt.title("Strip Plot")

plt.show()
```

```python
sns.stripplot(x='sex',
              y='age',
              data=dataset,
              jitter=True)

plt.title("Strip Plot with Jitter")

plt.show()
```

```python
sns.stripplot(x='sex',
              y='age',
              data=dataset,
              jitter=True,
              hue='survived')

plt.title("Strip Plot with Survival")

plt.show()
```

```python
sns.swarmplot(x='sex',
              y='age',
              data=dataset)

plt.title("Swarm Plot")

plt.show()
```

```python
sns.swarmplot(x='sex',
              y='age',
              data=dataset,
              hue='survived')

plt.title("Swarm Plot with Survival")

plt.show()
```

```python
corr = dataset.corr(numeric_only=True)

corr
```

```python
sns.heatmap(corr)

plt.title("Heatmap")

plt.show()
```

```python
sns.heatmap(corr,
            annot=True)

plt.title("Annotated Heatmap")

plt.show()
```

```python
sns.clustermap(corr)

plt.show()
```

```python
sns.histplot(dataset['fare'],
             kde=False,
             bins=10)

plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.title("Distribution of Fare")

plt.show()
```
