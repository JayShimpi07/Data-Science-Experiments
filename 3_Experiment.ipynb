Summary Statistics Grouped by Species

```python
df.groupby('species').agg({
    'sepal_length': ['mean', 'median', 'min', 'max', 'std'],
    'sepal_width': ['mean', 'median', 'min', 'max', 'std'],
    'petal_length': ['mean', 'median', 'min', 'max', 'std'],
    'petal_width': ['mean', 'median', 'min', 'max', 'std']
})
```

```python
df['species'].unique()
```

```python
from sklearn import preprocessing

enc = preprocessing.OneHotEncoder()

enc_df = pd.DataFrame(
    enc.fit_transform(df[['species']]).toarray()
)

enc_df
```

```python
df['sepal_length'].plot(kind='hist')

plt.xlabel("Sepal Length")

plt.title("Histogram of Sepal Length")

plt.show()
```
