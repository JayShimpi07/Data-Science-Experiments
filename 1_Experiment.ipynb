Data Formatting

```python
# Convert Species Column into Category Type

dataset['species'] = dataset['species'].astype('category')

dataset.dtypes
```

Data Normalization

```python
# Select Numerical Columns

x = dataset[['sepal_length',
             'sepal_width',
             'petal_length',
             'petal_width']]
```

```python
# Apply Min-Max Scaling

scaler = MinMaxScaler()

x_scaled = scaler.fit_transform(x)

x_scaled
```

```python
# Convert Scaled Data into DataFrame

normalized_dataset = pd.DataFrame(
    x_scaled,
    columns=['sepal_length',
             'sepal_width',
             'petal_length',
             'petal_width']
)

normalized_dataset.head()
```

Label Encoding

```python
# Display Unique Species

dataset['species'].unique()
```

```python
# Apply Label Encoding

label_encoder = LabelEncoder()

dataset['species_Label'] = label_encoder.fit_transform(dataset['species'])

dataset.head()
```

```python
# Encoded Values

dataset['species_Label'].unique()
```

One Hot Encoding

```python
# Apply One Hot Encoding

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(dataset[['species']])

encoded
```

```python
# Convert Encoded Data into DataFrame

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(['species'])
)

encoded_df.head()
```

Dummy Variable Encoding

```python
# Apply Dummy Encoding

dummy_dataset = pd.get_dummies(dataset,
                               columns=['species'],
                               drop_first=True)

dummy_dataset.head()
```

Simple Visualization

```python
# Scatter Plot

plt.scatter(dataset['sepal_length'],
            dataset['petal_length'])

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.title('Sepal Length vs Petal Length')

plt.show()
```

Correlation Heatmap

```python
# Correlation Heatmap

sns.heatmap(dataset.select_dtypes(include=np.number).corr(), # Select only numerical columns for correlation
            annot=True) # Annotate the heatmap with correlation values

plt.show()
```
