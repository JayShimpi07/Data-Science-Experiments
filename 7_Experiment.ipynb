Import libraries

```python
import nltk # Natural Language Toolkit
import numpy as np
import pandas as pd
```

Download NLTK data

```python
nltk.download('punkt_tab') # Download the Punkt tokenizer models
nltk.download('punkt') # Download the Punkt tokenizer models (for sentence splitting and word tokenization)
nltk.download('stopwords') # Download the stopwords corpus (for filtering out common words)
nltk.download('wordnet') # Download the WordNet corpus (for lemmatization and synonym lookup)
nltk.download('averaged_perceptron_tagger_eng') # Download the averaged perceptron tagger (for part-of-speech tagging)
```

Input text document

```python
text = """Text analytics is the process of analyzing unstructured text data.
Natural language processing helps computers understand human language.
Machine learning models use text analytics for sentiment analysis."""
```

Tokenization

```python
from nltk.tokenize import word_tokenize, sent_tokenize

sentences = sent_tokenize(text) # Tokenize the text into sentences
words = word_tokenize(text) # Tokenize the text into words (including punctuation)

print("Sentences:\n", sentences)
print("\nWords:\n", words)
```

Stopword removal

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english')) # Filter out stop words from the list of words

filtered_words = [w for w in words if w.lower() not in stop_words] # Create a list of words that are not stop words (case-insensitive)

print(filtered_words)
```

Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer() # Create a PorterStemmer object to perform stemming (reducing words to their root form)

stemmed_words = [stemmer.stem(w) for w in filtered_words] # Create a list of stemmed words by applying the stemmer to each filtered word

print(stemmed_words)
```

Lemmatization

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer() # Create a WordNetLemmatizer object to perform lemmatization (reducing words to their base form based on context)

lemmatized_words = [lemmatizer.lemmatize(w) for w in filtered_words] # Create a list of lemmatized words by applying the lemmatizer to each filtered word (using default 'n' for noun as the part of speech)

print(lemmatized_words)
```

POS Tagging

```python
pos_tags = nltk.pos_tag(filtered_words) # Perform part-of-speech tagging on the filtered words to identify their grammatical roles (e.g., noun, verb, adjective)

print(pos_tags)
```

TF-IDF Representation

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [ # A list of text documents to analyze
    "Text analytics is important",
    "Machine learning uses text analytics"
]

vectorizer = TfidfVectorizer() # Create a TfidfVectorizer object to convert the text documents into a TF-IDF matrix (term frequency-inverse document frequency)

tfidf_matrix = vectorizer.fit_transform(documents) # Fit the vectorizer to the documents and transform them into a TF-IDF matrix, where each row represents a document and each column represents a term (word) with its corresponding TF-IDF score

print("Feature Names:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```
